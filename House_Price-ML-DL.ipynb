## 4.Split the dataset into training and testing sets using train_test_split with a fixed random_state.


```python
from sklearn.model_selection import train_test_split
train_set,test_set = train_test_split(df,test_size=0.2,random_state=42)
```

```python
# feature
feature = train_set[['bedrooms','bathrms','stories','driveway','recroom','fullbase','gashw','airco','garagepl','prefarea']]
labels = train_set['price']
```

## 5.Train a Linear Regression model using LinearRegression() and fit it using training data.


```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
```

```python
onehot = OneHotEncoder()
```

```python
feature = onehot.fit_transform(feature)
```

```python
model = LinearRegression()
```

```python
model.fit(feature,labels)
```

```python
feature_test = test_set[['bedrooms','bathrms','stories','driveway','recroom','fullbase','gashw','airco','garagepl','prefarea']]

feature_test = onehot.transform(feature_test)
```

## 6.Predict on test data and create a small comparison table showing actual vs predicted results.


```python
prediction = model.predict(feature_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Assuming 'label_test' is true prices and 'prediction' are predicted prices
# label_test = test_set[['price']]

# 1. Mean Absolute Error (Average error in dollars)
mae = mean_absolute_error(label_test, prediction)
print(f"Mean Absolute Error: {mae}")

# 2. Root Mean Squared Error (Punishes large errors)
rmse = np.sqrt(mean_squared_error(label_test, prediction))
print(f"RMSE: {rmse}")
```
